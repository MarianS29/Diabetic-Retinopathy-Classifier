```python
import os
import shutil
from tqdm import tqdm

# ==========================================
# CONFIGURARE CĂI
# ==========================================
# Căile către cele două dataset-uri pe care vrei să le combini
DATASET_1 = "../datasets/aptos_ben_graham"
DATASET_2 = "../datasets/Diabetic_Balanced_Data" # Asigură-te că folderul se numește exact așa

# Folderul final în care vor fi vărsate ambele seturi
OUTPUT_DIR = "../datasets/dataset_combinat"

# ==========================================
# LOGICA DE CONCATENARE
# ==========================================
def copy_and_prefix_dataset(src_dataset_dir, prefix):
    """
    Parcurge structura split/clasă dintr-un dataset sursă și copiază imaginile
    în folderul destinație adăugând un prefix numelui fișierului.
    """
    if not os.path.exists(src_dataset_dir):
        print(f"⚠️ Eroare: Nu găsesc folderul dataset-ului '{src_dataset_dir}'!")
        return

    # Presupunem structura standard: train/val/test
    splits = ['train', 'val', 'test']
    clase = ['0', '1', '2', '3', '4']

    for split in splits:
        for c in clase:
            src_path = os.path.join(src_dataset_dir, split, c)
            dst_path = os.path.join(OUTPUT_DIR, split, c)

            # Dacă folderul sursă nu există (ex: poate Dataset 2 nu are 'val'), trecem mai departe
            if not os.path.exists(src_path):
                continue

            # Ne asigurăm că folderul destinație există
            os.makedirs(dst_path, exist_ok=True)

            # Găsim toate imaginile
            images = [f for f in os.listdir(src_path) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
            
            if not images:
                continue

            # Le copiem cu un Progress Bar vizual
            for img_name in tqdm(images, desc=f"Mutare {prefix.strip('_')} -> {split.upper()}/Clasa {c}", leave=False):
                src_img_file = os.path.join(src_path, img_name)
                
                # Creăm noul nume ca să evităm suprascrierea (ex: "aptos_img_001.png")
                new_img_name = f"{prefix}{img_name}"
                dst_img_file = os.path.join(dst_path, new_img_name)
                
                shutil.copy2(src_img_file, dst_img_file)

def main():
    print(f"🚀 START: Concatenare '{DATASET_1}' și '{DATASET_2}'\n")
    print(f"Folder destinație: '{OUTPUT_DIR}'\n")
    
    # 1. Copiem primul set de date (îi punem prefixul "aptos_")
    print(f"📂 Procesăm primul dataset: {DATASET_1}")
    copy_and_prefix_dataset(DATASET_1, prefix="aptos_")
    print("✅ Primul dataset a fost copiat cu succes.\n")
    
    # 2. Copiem al doilea set de date (îi punem prefixul "bal_")
    print(f"📂 Procesăm al doilea dataset: {DATASET_2}")
    copy_and_prefix_dataset(DATASET_2, prefix="bal_")
    print("✅ Al doilea dataset a fost copiat cu succes.\n")
    
    print(f"🎉 OPERAȚIUNE ÎNCHEIATĂ! Ai ambele dataset-uri combinate în '{OUTPUT_DIR}'.")
    
    # Afișăm o scurtă statistică a folderului final de antrenament
    print("\n📊 Statistică nouă pentru TRAIN:")
    for c in ['0', '1', '2', '3', '4']:
        train_class_dir = os.path.join(OUTPUT_DIR, 'train', c)
        if os.path.exists(train_class_dir):
            count = len([f for f in os.listdir(train_class_dir) if f.lower().endswith(('.png', '.jpg', '.jpeg'))])
            print(f"   ► Clasa {c}: {count} imagini totale")

if __name__ == '__main__':
    main()
```
