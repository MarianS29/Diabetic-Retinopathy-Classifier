# Ben Graham preprocessing

Ruleaza acelasi pipeline ca `ben_graham_preprocess_dataset.py`: scale radius, normalizare cu blur Gaussian, masca circulara si conversie grayscale RGB pentru compatibilitate vizuala cu `Diabetic_Balanced_Data`.

```python
from argparse import Namespace
from pathlib import Path

from ben_graham_preprocess_dataset import process_dataset, print_counts

cwd = Path.cwd()
ROOT = cwd if (cwd / "datasets").exists() else cwd.parent

args = Namespace(
    input_dir=ROOT / "datasets" / "aptos" / "aptos_augmented_balanced",
    output_dir=ROOT / "datasets" / "aptos" / "aptos_ben_graham_matched",
    output_size=512,
    radius=300,
    black_threshold=7,
    mask_ratio=1.0,
    grayscale=True,
    contrast=8.0,
    overwrite=True,
    jpeg_quality=95,
)

processed_count = process_dataset(args)
print(f"\nProcesate {processed_count} imagini.")
print_counts(args.output_dir)
```
