## Partea 3: Soluționarea Dezechilibrului de Clase (pos_weight)
În mediul clinic real, avem mult mai mulți pacienți sănătoși decât bolnavi. În acest experiment, forțăm rețeaua să penalizeze mult mai sever o greșeală de tip "Fals Negativ" (adică să trimită acasă un pacient bolnav spunându-i că e sănătos).

* `pos_weight 3.0` = O greșeală pe pacient bolnav costă cât 3 greșeli pe pacienți sănătoși.
* `pos_weight 5.0` = O greșeală pe pacient bolnav costă de 5 ori mai mult.

```python
# 1. BCE Standard (Fără ponderare, Baseline - deja rulat)
# %run train_det.py --dataset aptos --model resnet50 --loss bce --lr 0.0001 --epochs 15 --num_workers 0

# 2. BCE cu penalizare Moderată (x3)
%run train_det.py --dataset aptos --model resnet50 --loss bce --pos_weight 3.0 --lr 0.0001 --epochs 15 --batch_size 32 --num_workers 8
```

```python
# 3. BCE cu penalizare Severă (x5) - Observăm dacă modelul devine prea precaut (dă prea multe alarme false)
%run train_det.py --dataset aptos --model resnet50 --loss bce --pos_weight 5.0 --lr 0.0001 --epochs 15 --batch_size 32 --num_workers 8
```
